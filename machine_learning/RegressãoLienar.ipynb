```python
from sklearn.linear_model import LinearRegression


# Orçamento de Marketing
x = [[1000],[2500],[3250],[5000],[6500]]
# Lucro em Vendas
y = [10_000, 20_500, 33_250, 59_000, 77_000]


modelo = LinearRegression()


# Treinando o modelo
modelo.fit(x,y)


predicao = modelo.predict([[5000]])


print(f"Predição de lucro: R$ {predicao[0]:,.2f}")



```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Orçamento de Marketing
x = [[1000], [2500], [3250], [5000], [6500]]

# Lucro em Vendas
y = [10000, 20500, 33250, 59000, 77000]

# Criar e treinar o modelo
modelo = LinearRegression()
modelo.fit(x, y)

# Fazer a previsão
previsao = modelo.predict([[20000]])

# Criar o gráfico
plt.scatter([1000, 2500, 3250, 5000, 6500], y,
            color="blue", label="Lucro real")

plt.plot([1000, 2500, 3250, 5000, 6500],
         modelo.predict(x),
         color="red", label="Regressão linear")

plt.title("Orçamento de Marketing x Lucro em Vendas")
plt.xlabel("Orçamento de Marketing (R$)")
plt.ylabel("Lucro em Vendas (R$)")

plt.legend()
plt.grid(True)
plt.show()

# Mostrar o resultado
print(f"Previsão de lucro: R$ {previsao[0]:,.2f}")

```
